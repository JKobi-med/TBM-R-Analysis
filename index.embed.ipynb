## Study design

In this activly controlled, randomzied, open- label CGPT: In this two-site, longitudinal, randomized, single-blind intervention study, retired healthy older adults received weekly piano training (intervention) or active music sensitization (active control) for 12 months to assess cognitive, perceptual-motor, and brain plasticity outcomes. @jamesTrainBrainMusic2020a

Blood samples were taken at three different time points. The first (T0) one week before the weekly cycle of practical or theoretical music lessons were started. The second (T1) after 6 months and the third (T2) again after 6 months so one year after T0.

Because multiple tests were performed on each inidivuals time point (not only blood samples but also fMRI as part of the original study) T0 fell between March and June 2019, T1 between August 2019 and February 2020 and T2 between June 2020 and November 2020.

## Measurement of DNA methylation

### Target Genes

We focused on the primer regions of three genes commonly neuroplasticity-associated and hypthecised being potentially epigenetically modifiable by playing the piano: BDNF4, PSD95, MAPT (Tau).

### DNA isolation

Genomic DNA (gDNA) was isolated at the Institute of Molecular and Translational Therapeutic Strategies of Hannover Medical School from 50 µL blood using the “NucleoMag Blood 200 µl” kit (MACHEREY-NAGEL GmbH & Co. KG, Düren, Deutschland) in concordance with the instructions of the manufacturer. The isolated gDNA was stored at -20 °C. The blood samples have been received anonymized and prior to gDNA isolation. Then the samples were additionaly randomized *(by doing what?)* to minimize potential batch effects. DNA samples were diluted in 96-well plates to a fixed concentration of 10 ng/µl.

### Bisulfite Conversion

The gDNA of all samples was then bisullfite converted via the “EZ-96 DNA Methylation-Lightning™ MagPrep” kit (Zymo Research Corporation, Irvine, CA, USA) also following the manufacturers instructions.

### Primer design

With Ensembl and Geneious Pro 5.6.7 the primer regions of our target genes were located and the sequences of the primer regions analyzed. Aided by Biocalculator (metabion), NetPrimer (Premier Biosoft), Methyl Primer Express v1.0 (Applied Biosystems) and Bisulfite Primer Seeker (Zymo) we designed primers for PCR with bisulfite-converted DNA (see @tbl-primers-pcr) and sequencing primers (@tbl-primers-seq) - later used to prepare Sanger Sequencing - and orderd these primers from metabion GmbH (Martinsried, Deutschland).

### Gene Amplification

The primers were dissolved in nuclease-free water. The proportions have been predetermined by the manufacturer. The stock solution 100 pmol. It was then used to derive a working solution of 20 pmol (1:5). Both solutions were prexserved atg -20 °C. Amplification was then performed together with HotStarTaq® PCR Master Mix Kit (Qiagen, Hilden, Deutschland). The samples were then puriefed with the PCR purification kit “CleanNGS” (CleanNA, Waddinxveen, Niederlande).

### Sanger Sequencing

The purfied PCR prodcus were sequenced with the BigDye Terminator v3.1 Cycle Sequencing Kit (Applied Biosystems, Foster City, CA, USA) and th e sequencing primers mentioned above (@tbl-primers-seq). The sequencing product were purifed with the “CleanDTR” kit (CleanNA, Waddinxveen, Niederlande).

Then using the Applied Biosystems’ 3500 Genetic Analyzer (ABI 3500xL) Capillary electrophoresis was performed.

### Data Processing via ESME

The data derived from the ABI 3500xL was then processed with Epigenetic Sequencing Methylation (ESME) analysis software.


    Loading required package: Matrix
    Attaching package: 'Matrix'The following objects are masked from 'package:tidyr':

        expand, pack, unpackWelcome to emmeans.
    Caution: You lose important information if you filter this package's results.
    See '? untidy'boundary (singular) fit: see help('isSingular')Warning in ggsignif::geom_signif(data = pw, aes(xmin = x1, xmax = x2,
    annotations = label, : Ignoring unknown aesthetics: xmin, xmax, annotations,
    and y_positionWarning in ggsignif::geom_signif(data = pw, aes(xmin = x1, xmax = x2,
    annotations = label, : Ignoring unknown aesthetics: xmin, xmax, annotations,
    and y_positionboundary (singular) fit: see help('isSingular')Warning in ggsignif::geom_signif(data = pw, aes(xmin = x1, xmax = x2,
    annotations = label, : Ignoring unknown aesthetics: xmin, xmax, annotations,
    and y_position



#### Time difference per group

Our main goal we aimed to determine whether the degree of methylation differed between the intervention (piano lessons) and control groups.

-   Bei BDNF zeigt sich eine leichte Differenz zwischen der Kontrollgruppe (Median: 0.07, Mean: 0.07326) und der Pianogruppe (Median: 0.064, Mean: 0.06598), welche eine niedrigere Metyhlierung (Median: -0.006, Mean: -0.00728) aufwies. **Streuungsmaße?**
-   Weiterhin zeigen sich für die arithmetischen Mittel von BDNF und PSD95 zwei Modi.
-   Tau war unimodal.

``` python
library(dplyr)
library(tidyr)
library(ggplot2)
library(patchwork)
library(lme4)
library(emmeans)

p_to_star <- function(p){
  ifelse(is.na(p), "NA",
  ifelse(p < 0.001, "***",
  ifelse(p < 0.01,  "**",
  ifelse(p < 0.05,  "*", "ns"))))
}

make_interaction_plot_sig <- function(df_wide, title,
                                      ylab = "Mean methylation (mean across CpGs)",
                                      control_label = "Controll",
                                      piano_label = "Piano") {

  # wide -> long
  df_long <- df_wide %>%
    pivot_longer(cols = c(T0, T1, T2),
                 names_to = "Zeitpunkt",
                 values_to = "Mean") %>%
    mutate(
      Zeitpunkt = factor(Zeitpunkt, levels = c("T0", "T1", "T2")),
      Gruppe = factor(Gruppe, levels = c(control_label, piano_label)),
      ID = factor(ID)
    ) %>%
    filter(!is.na(Mean))

  # mixed model (participant-level)
  fit <- lmer(Mean ~ Gruppe * Zeitpunkt + (1 | ID), data = df_long, REML = TRUE)

  # Piano vs Control at each timepoint
  emm <- emmeans(fit, ~ Gruppe | Zeitpunkt)
  contr <- as.data.frame(contrast(emm, method = list("Piano - Control" = c(-1, 1))))

  pvals <- contr %>%
    transmute(
      Zeitpunkt,
      p.value,
      label = p_to_star(p.value)
    )

  # summary stats for plot (95% CI)
  summary_stats <- df_long %>%
    group_by(Zeitpunkt, Gruppe) %>%
    summarise(
      n = n(),
      mean = mean(Mean),
      sd = sd(Mean),
      se = sd / sqrt(n),
      ci = qt(0.975, df = n - 1) * se,
      .groups = "drop"
    )

  # y-positions for labels (above the top CI per timepoint)
  y_pad <- 0.06 * diff(range(summary_stats$mean, na.rm = TRUE))
  ann <- summary_stats %>%
    group_by(Zeitpunkt) %>%
    summarise(y = max(mean + ci, na.rm = TRUE) + y_pad, .groups = "drop") %>%
    left_join(pvals, by = "Zeitpunkt")

  ggplot(summary_stats, aes(x = Zeitpunkt, y = mean, group = Gruppe, colour = Gruppe)) +
    geom_line() +
    geom_point() +
    geom_errorbar(aes(ymin = mean - ci, ymax = mean + ci), width = 0.2) +
    geom_text(data = ann, aes(x = Zeitpunkt, y = y, label = label),
              inherit.aes = FALSE, vjust = 0, size = 4) +
    theme_classic() +
    labs(
      title = title,
      x = "Time Point",
      y = ylab,
      caption = "Error bars are 95% CI. Stars: Piano vs Control within timepoint from mixed model."
    )
}

# BDNF
p_bdnf <- make_interaction_plot_sig(
  bdnf_mean_wide,
  title = "Interaction Plot of BDNF Methylation Levels by Group and Time"
)
```


    Loading required package: Matrix
    Attaching package: 'Matrix'The following objects are masked from 'package:tidyr':

        expand, pack, unpackWelcome to emmeans.
    Caution: You lose important information if you filter this package's results.
    See '? untidy'Warning in ggsignif::geom_signif(data = pw, aes(xmin = x1, xmax = x2,
    annotations = label, : Ignoring unknown aesthetics: xmin, xmax, annotations,
    and y_positionWarning in ggsignif::geom_signif(data = pw, aes(xmin = x1, xmax = x2,
    annotations = label, : Ignoring unknown aesthetics: xmin, xmax, annotations,
    and y_positionWarning in ggsignif::geom_signif(data = pw, aes(xmin = x1, xmax = x2,
    annotations = label, : Ignoring unknown aesthetics: xmin, xmax, annotations,
    and y_position



#### Group

``` python
library(dplyr)
library(tidyr)
library(ggplot2)
library(patchwork)
library(lme4)
library(emmeans)

p_to_star <- function(p){
  ifelse(is.na(p), "NA",
  ifelse(p < 0.001, "***",
  ifelse(p < 0.01,  "**",
  ifelse(p < 0.05,  "*", "ns"))))
}

plot_gene_piano_vs_control <- function(df_wide, gene_title,
                                       ylab = "Mean methylation (across CpGs)") {

  # Ensure factor order so the contrast is Piano - Control
  df_long <- df_wide %>%
    pivot_longer(cols = c(T0, T1, T2),
                 names_to = "Zeitpunkt",
                 values_to = "Mean") %>%
    mutate(
      Zeitpunkt = factor(Zeitpunkt, levels = c("T0", "T1", "T2")),
      Gruppe = factor(Gruppe, levels = c("Controll", "Piano")),
      ID = factor(ID)
    ) %>%
    filter(!is.na(Mean))

  # Mixed model (participant-level)
  fit <- lmer(Mean ~ Gruppe * Zeitpunkt + (1 | ID), data = df_long, REML = TRUE)

  # Piano vs Control at each timepoint
  emm <- emmeans(fit, ~ Gruppe | Zeitpunkt)
  contr <- as.data.frame(contrast(emm, method = list("Piano - Control" = c(-1, 1))))

  pvals <- contr %>%
    transmute(
      Zeitpunkt,
      p.value,
      label = p_to_star(p.value)
    )

  # Plot stats (95% CI)
  summary_stats <- df_long %>%
    group_by(Zeitpunkt, Gruppe) %>%
    summarise(
      n = n(),
      mean = mean(Mean),
      sd = sd(Mean),
      se = sd / sqrt(n),
      ci = qt(0.975, df = n - 1) * se,
      .groups = "drop"
    )

  # Annotation y-positions (above the top CI at each timepoint)
  y_pad <- 0.05 * diff(range(summary_stats$mean, na.rm = TRUE))
  ann <- summary_stats %>%
    group_by(Zeitpunkt) %>%
    summarise(y = max(mean + ci, na.rm = TRUE) + y_pad, .groups = "drop") %>%
    left_join(pvals, by = "Zeitpunkt")

  ggplot(summary_stats, aes(x = Zeitpunkt, y = mean, group = Gruppe, colour = Gruppe)) +
    geom_line() +
    geom_point() +
    geom_errorbar(aes(ymin = mean - ci, ymax = mean + ci), width = 0.2) +
    geom_text(data = ann, aes(x = Zeitpunkt, y = y, label = label),
              inherit.aes = FALSE, vjust = 0, size = 4) +
    theme_classic() +
    labs(
      title = gene_title,
      x = "Time Point",
      y = ylab,
      caption = "Error bars are 95% CI. Stars: Piano vs Control within timepoint from mixed model."
    )
}

p_bdnf  <- plot_gene_piano_vs_control(bdnf_mean_wide,  "BDNF")
p_tau   <- plot_gene_piano_vs_control(tau_mean_wide,   "Tau")
p_psd95 <- plot_gene_piano_vs_control(psd95_mean_wide, "PSD95")

(p_bdnf / p_tau / p_psd95) + plot_layout(guides = "collect") &
  theme(legend.position = "right")
```
